# Prompt Evaluation — Part 1: Building a Dataset

**Why evaluate?** "It looks good to me" doesn't scale. To improve a prompt with confidence you need a repeatable measurement. Every eval has three moving parts, and we build them up across these four notebooks:

1. **A dataset** of test cases (this notebook)
2. **A way to run** each test case through your prompt
3. **A way to grade** the output — the hard part

**Key takeaways from this notebook:**

- **Bootstrap the dataset with Claude itself.** Hand-writing dozens of test cases is slow. Instead, describe the *shape* of task you care about (here: small AWS-related Python/JSON/Regex snippets) and let the model generate the cases. You can always hand-edit afterwards.
- **The generation prompt gives an explicit example of the output shape** (the ` ```json ` block). Showing the structure you want is far more reliable than describing it in prose.
- Aim for test cases that are **diverse and representative** of real usage — that's what makes the eval score meaningful.
