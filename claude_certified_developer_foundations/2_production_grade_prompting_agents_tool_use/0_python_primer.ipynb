## 0. How a notebook works

- A notebook is a list of **cells**. A code cell runs when you press **Shift+Enter**.
- `[ ]` = never run · `[*]` = running · `[3]` = the 3rd cell you ran.
- **Order matters.** Cells share one memory ("the kernel"), so a variable made in one cell
  is still there in the next.
- The **last expression** in a cell is displayed automatically — no `print()` needed.

```python
greeting = "hello"     # store a value under a name
greeting                # last line of a cell -> displayed automatically
```

```python
# The variable is still alive in this cell — same kernel.
print(greeting)
print("shouted:", greeting.upper())
```

## 1. Variables and naming

- `name = value`. No type declaration, no `let`/`var`.
- `lower_snake_case` for normal variables.
- `UPPER_SNAKE_CASE` by convention means "a constant — set once, don't change".
  That's why the real notebook has `PUZZLE`, `EASY`, `INPUT_PER_MTOK`.
- `#` starts a comment.

```python
model = "claude-opus-5"          # normal variable
INPUT_PER_MTOK = 5.00            # constant by convention (Python won't stop you changing it)
max_tokens = 4000                # whole number (int)
seconds = 1.5                    # decimal (float)
ready = True                     # boolean: True / False (capitalised!)

print(model, INPUT_PER_MTOK, max_tokens, seconds, ready)
```

## 2. Strings

Three things the real notebook relies on:

1. `"..."` and `'...'` are identical. Use whichever avoids escaping.
2. `"""..."""` spans multiple lines.
3. **Adjacent strings glue together automatically** — that's how `PUZZLE` is built across
   three lines with no `+`.

```python
single = 'one way'
double = "the same thing"

multi = """first line
second line"""

# Implicit concatenation: no +, just adjacent strings inside brackets.
PUZZLE = (
    "A shop sells apples in bags of 6 and 11. "
    "What is the largest number you cannot buy? "
    "Give one sentence."
)

print(multi)
print()
print(PUZZLE)
```

## 3. f-strings — putting values inside text

Put `f` before the quote, then `{expression}` inside. Used for every `print` in the real
notebook.

```python
effort = "max"
tokens = 549

print(f"effort was {effort} and it used {tokens} tokens")

# Any expression works inside the braces, not just a name:
print(f"double that is {tokens * 2}")
```

### Format specifiers — the `:>6` and `:.4f` bits

After a colon inside the braces you can control **width and precision**. This is what makes
the effort-sweep output line up in neat columns.

| Spec | Means |
|---|---|
| `:>5` | right-align in 5 characters wide |
| `:<5` | left-align |
| `:.4f` | a decimal number, 4 places after the point |
| `!r` | show the value the way Python would *write* it (quotes included) |

```python
for name, value in [("low", 6), ("high", 238), ("max", 549)]:
    print(f"{name:>5}  {value:>6} tokens")   # columns line up

print()
cost = 0.0062431
print(f"unformatted: {cost}")
print(f"4 decimals:  {cost:.4f}")

print()
empty = ""
print(f"without !r the empty string is invisible: {empty}")
print(f"with !r you can see it:                   {empty!r}")
```

`!r` is why the real notebook can *show* you that thinking text is an empty string
(`''`) rather than printing nothing at all.

## 4. Lists — an ordered sequence

- Written with `[ ]`, items separated by commas.
- **Counting starts at 0.**
- `.append(x)` adds to the end.
- `[:3]` takes a **slice** — the first 3 items. On a string it takes the first 3 characters.

```python
efforts = ["low", "medium", "high", "xhigh", "max"]

print("all:      ", efforts)
print("first:    ", efforts[0])       # 0, not 1
print("last:     ", efforts[-1])      # negative counts from the end
print("first two:", efforts[:2])      # a slice
print("how many: ", len(efforts))

efforts.append("extreme")             # modifies the list in place
print("after append:", efforts)

answer = "Canberra is the capital of Australia"
print("first 8 chars:", answer[:8])   # slicing a string
```

## 5. Dictionaries — labelled values

- Written with `{ }` as `key: value` pairs.
- Look a value up by **key**, not position: `d["key"]`.
- `.get("key", fallback)` returns the fallback instead of crashing when the key is missing.

Every API argument in the real notebook is a dictionary — `{"type": "adaptive"}`,
`{"effort": "high"}`, `{"role": "user", "content": ...}`.

```python
thinking = {"type": "adaptive", "display": "summarized"}

print(thinking)
print("type is:", thinking["type"])

# Missing key with [] would crash. .get() gives a fallback instead:
print("effort key:", thinking.get("effort", "(not set)"))

READINGS = {"Sydney": 22.5, "Melbourne": 18.0}
print("Sydney:", READINGS["Sydney"])
print("Perth: ", READINGS.get("Perth", "unknown"))   # exactly the notebook's tool-result line
```

## 6. Nesting — a list of dictionaries

This is the `messages` shape. Read it outside-in: a **list**, containing one **dictionary**,
whose `content` is a string.

```python
messages = [
    {"role": "user", "content": "What is 2 + 2?"},
]

print("the list has", len(messages), "item")
print("first item: ", messages[0])
print("its role:   ", messages[0]["role"])

# A second turn is just another dictionary appended to the list.
messages.append({"role": "assistant", "content": "4"})
for m in messages:
    print(f"  {m['role']:>9}: {m['content']}")
```

⚠️ Note `{m['role']}` uses **single** quotes inside a **double**-quoted f-string. Reusing the
same quote type would end the string early.

## 7. Objects and attributes — dots vs. brackets

This trips up everyone. Two different things that both hold data:

| | Access with | Example |
|---|---|---|
| **Dictionary** (data you build) | `["key"]` | `thinking["type"]` |
| **Object** (things the SDK returns) | `.attribute` | `response.usage.output_tokens` |

The API response is an **object**, so you write `block.type`, not `block["type"]`.
Dots can chain: `response.usage.output_tokens` means "the response's usage, then its
output_tokens".

Below we build fake stand-ins with the same shape as real API responses, so the rest of this
notebook runs with no API key.

```python
class Block:
    """Stand-in for a content block from the API."""
    def __init__(self, **fields):
        self.__dict__.update(fields)     # turn each keyword into an attribute
    def __repr__(self):
        return f"<{self.type}>"

class Usage:
    def __init__(self, input_tokens, output_tokens):
        self.input_tokens = input_tokens
        self.output_tokens = output_tokens

class Response:
    def __init__(self, content, usage):
        self.content = content
        self.usage = usage

# Shaped exactly like a real reply with thinking turned on.
response = Response(
    content=[
        Block(type="thinking", thinking="Let me work through the bags of 6 and 11...",
              signature="abc123signature"),
        Block(type="text", text="The answer is 49."),
    ],
    usage=Usage(input_tokens=56, output_tokens=238),
)

print("content is a list:", response.content)
print("first block type: ", response.content[0].type)      # dot, not bracket
print("chained access:   ", response.usage.output_tokens)
```

## 8. `for` loops

`for X in Y:` repeats once per item. The **indented** lines are the body — indentation is
how Python marks blocks, not `{ }`.

```python
for block in response.content:
    print("this block is a", block.type)

print()
for effort in ["low", "high"]:
    print(f"  running at {effort}")
```

## 9. `if` / `elif` / `else`

`==` tests equality (`=` assigns). This is the pattern for handling each block type.

```python
for block in response.content:
    if block.type == "thinking":
        print("THINKING:", block.thinking)
    elif block.type == "text":
        print("ANSWER:  ", block.text)
    else:
        print("something else:", block.type)
```

## 10. List comprehensions — build a list in one line

`[ WHAT_TO_KEEP for ITEM in SEQUENCE ]`, with an optional `if` filter on the end.

The real notebook uses `[block.type for block in response.content]` to show block order.

```python
# The long way:
types = []
for block in response.content:
    types.append(block.type)
print("long way: ", types)

# The same thing as a comprehension:
types = [block.type for block in response.content]
print("short way:", types)

# With a filter on the end:
only_text = [b.text for b in response.content if b.type == "text"]
print("filtered: ", only_text)
```

## 11. Generators and `next()` — the "find the first match" idiom

Swap the `[ ]` for `( )` and you get a **generator**: it produces items one at a time,
on demand, instead of building the whole list.

`next(...)` pulls **one** item out — the first match — then stops looking.

```python
# A generator doesn't compute anything yet:
gen = (b for b in response.content if b.type == "thinking")
print("the generator itself:", gen)

# next() pulls the first match out of it:
print("first match:", next(gen))
```

```python
# Written as one line — this is the exact line from the extended thinking notebook:
thinking_block = next(b for b in response.content if b.type == "thinking")

print("found:    ", thinking_block)
print("its text: ", thinking_block.thinking)
print("signature:", thinking_block.signature)
```

**Why not `response.content[0]`?** Because position isn't guaranteed. Selecting by
`.type` says what you actually mean and survives a changed block order.

**Why a generator, not a list?** `next()` stops at the first hit. A list comprehension would
build every match first and then throw all but one away.

### ⚠️ `next()` crashes when nothing matches

Add a **second argument** and you get that value instead of a crash.

Note the extra parentheses: they're optional when the generator is the only argument, and
**required** once you add a fallback.

```python
no_thinking = [Block(type="text", text="just an answer")]

# Without a fallback -> crash
try:
    next(b for b in no_thinking if b.type == "thinking")
except StopIteration:
    print("crashed with StopIteration")

# With a fallback -> None. Note the (extra) parentheses around the generator.
found = next((b for b in no_thinking if b.type == "thinking"), None)
print("with fallback:", found)

# The notebook's measure() uses "" as the fallback for missing text:
answer = next((b.text for b in no_thinking if b.type == "text"), "")
print("answer:", repr(answer))
```

## 12. Functions

- `def name(inputs):` then an indented body.
- `return` hands a value back.
- The `"""..."""` first line is a **docstring** — a human description.
- `param=value` in the definition makes that input **optional**.

```python
def cost(usage, input_rate=5.00, output_rate=25.00):
    """Work out the dollar cost of one response."""
    return (usage.input_tokens * input_rate
            + usage.output_tokens * output_rate) / 1_000_000


print(cost(response.usage))                       # uses the default rates
print(cost(response.usage, output_rate=50.00))    # override just one
print(f"formatted: ${cost(response.usage):.4f}")
```

`1_000_000` is just `1000000` — underscores are ignored by Python and only there so humans
can read the number.

## 13. Keyword arguments and multi-line calls

`name=value` when calling a function labels each input, so order stops mattering. Every API
call in the real notebook is written this way, one argument per line, with a **trailing
comma** after the last one (legal in Python, and it keeps diffs clean).

```python
def describe(model, max_tokens, thinking=None):
    return f"{model} | max_tokens={max_tokens} | thinking={thinking}"


print(describe(
    model="claude-opus-5",
    max_tokens=4000,
    thinking={"type": "adaptive", "display": "summarized"},   # <- trailing comma is fine
))

# Order doesn't matter with keywords:
print(describe(max_tokens=100, model="claude-sonnet-5"))
```

## 14. `try` / `except` — handling an expected error

`try:` the risky thing; `except SomeError as e:` runs only if that error happened, with the
error itself available as `e`. The real notebook uses it to *demonstrate* the
`budget_tokens` 400 rather than let the notebook die.

```python
def risky(n):
    if n == 0:
        raise ValueError("n must not be zero")
    return 100 / n


for n in [4, 0]:
    try:
        print(f"100/{n} =", risky(n))
    except ValueError as e:
        print(f"caught an error: {e}")

print("the notebook keeps running")
```

## 15. Putting it together

Here is a real cell from `3_extended_thinking.ipynb`, annotated line by line — running
against our fake `response` so it needs no API key. Every idiom above appears in it.

```python
import time                                    # standard library module

def measure(effort, prompt):                   # 12. function, two inputs
    """Run one request at a given effort and report what it cost."""   # docstring
    started = time.time()                      # 1. variable

    # In the real notebook this is an API call. Here we reuse the fake response.
    result = response                          # 7. an object with attributes

    answer = next(                             # 11. first match, with a fallback
        (b.text for b in result.content if b.type == "text"),
        "",
    )
    return {                                   # 5. returns a dictionary
        "effort": effort,
        "output_tokens": result.usage.output_tokens,     # 7. chained attributes
        "seconds": round(time.time() - started, 1),
        "answer": answer.strip().replace("\n", " ")[:80],  # 4. slice to 80 chars
    }


# 10. comprehension: call measure() once per effort level
results = [measure(e, "the puzzle") for e in ["low", "high", "max"]]

for r in results:                              # 8. loop
    # 3. f-string with alignment specs
    print(f"{r['effort']:>5}  {r['output_tokens']:>6} out tok  {r['seconds']:>5}s  {r['answer']}")
```
