## 0. How a notebook works

- A notebook is a list of **cells**. A code cell runs when you press **Shift+Enter**.
- `[ ]` = never run · `[*]` = running · `[3]` = the 3rd cell you ran.
- **Order matters.** Cells share one memory ("the kernel"), so a variable made in one cell
  is still there in the next.
- The **last expression** in a cell is displayed automatically — no `print()` needed.

````python
greeting = "hello"     # store a value under a name
greeting                # last line of a cell -> displayed automatically
````

````python
# The variable is still alive in this cell — same kernel.
print(greeting)
print("shouted:", greeting.upper())
````

- `lower_snake_case` for normal variables.
- `UPPER_SNAKE_CASE` by convention means "a constant — set once, don't change".
  That's why the notebooks have `PUZZLE`, `DATASET`, `GET_BALANCE`.
- `#` starts a comment.

````python
model = "claude-opus-5"          # normal variable
INPUT_PER_MTOK = 5.00            # constant by convention (Python won't stop you changing it)
max_tokens = 4000                # whole number (int)
seconds = 1.5                    # decimal (float)
ready = True                     # boolean: True / False (capitalised!)

print(model, INPUT_PER_MTOK, max_tokens, seconds, ready)
````

## 2. Strings

1. `"..."` and `'...'` are identical. Use whichever avoids escaping.
2. `"""..."""` spans multiple lines.
3. **Adjacent strings glue together automatically** — that's how long prompts are built
   across several lines with no `+`.

````python
multi = """first line
second line"""

# Implicit concatenation: no +, just adjacent strings inside brackets.
PUZZLE = (
    "A shop sells apples in bags of 6 and 11. "
    "What is the largest number you cannot buy? "
    "Give one sentence."
)

print(multi)
print()
print(PUZZLE)
````

### String methods you'll meet

`.strip()` trims whitespace · `.lower()` / `.upper()` change case · `.replace(a, b)`
substitutes · `[:n]` slices.

`.lower()` on both sides is how you compare **case-insensitively** — used in the evals
notebook to check whether an owner's name appears in a summary.

````python
messy = "   BEN will roll back the deploy.\n"

print(repr(messy.strip()))
print(messy.strip().lower())
print(messy.strip().replace("BEN", "Ana"))
print("first 10 chars:", repr(messy.strip()[:10]))

# Case-insensitive containment check
print("ben appears:", "ben" in messy.lower())
````

## 3. f-strings — putting values inside text

Put `f` before the quote, then `{expression}` inside.

````python
effort = "max"
tokens = 549

print(f"effort was {effort} and it used {tokens} tokens")
print(f"double that is {tokens * 2}")      # any expression works inside the braces
````

### Format specifiers — the `:>6` and `:.4f` bits

| Spec | Means |
|---|---|
| `:>5` | right-align in 5 characters wide |
| `:<5` | left-align |
| `:.4f` | a decimal number, 4 places after the point |
| `!r` | show the value the way Python would *write* it (quotes included) |

````python
for name, value in [("low", 6), ("high", 238), ("max", 549)]:
    print(f"{name:>5}  {value:>6} tokens")   # columns line up

print()
cost = 0.0062431
print(f"unformatted: {cost}")
print(f"4 decimals:  {cost:.4f}")

print()
empty = ""
print(f"without !r the empty string is invisible: {empty}")
print(f"with !r you can see it:                   {empty!r}")
````

`!r` is why the extended thinking notebook can *show* you that thinking text is an empty
string (`''`) rather than printing nothing at all.

## 4. Lists — an ordered sequence

- `[ ]`, comma separated. **Counting starts at 0.** `-1` is the last item.
- `.append(x)` adds to the end. `[:3]` is a **slice**.

````python
efforts = ["low", "medium", "high", "xhigh", "max"]

print("first:    ", efforts[0])       # 0, not 1
print("last:     ", efforts[-1])      # negative counts from the end
print("first two:", efforts[:2])
print("how many: ", len(efforts))

efforts.append("extreme")
print("after append:", efforts)
````

## 5. Sets — unordered, no duplicates

`set(...)` throws away order and duplicates. Two uses in the notebooks:

- **Counting distinct values** — `len(set(outputs))` tells you how many *different* answers
  a prompt gave across runs.
- **Order-independent comparison** — comparing two JSON arrays as sets is exactly the
  grader trick in the evals notebook: same items, any order, still equal.

````python
runs = ["BILLING", "Billing", "BILLING", "billing"]
print("distinct outputs:", len(set(runs)), "->", set(runs))

# Order-independent equality — the code-grader trick
want = {"Canberra", "Wellington", "Suva"}
got  = {"Wellington", "Suva", "Canberra"}
print("same as a list? ", ["Canberra", "Wellington", "Suva"] == ["Wellington", "Suva", "Canberra"])
print("same as a set?  ", want == got)
````

## 6. Dictionaries — labelled values

- `{ }` as `key: value`. Look up by **key**: `d["key"]`.
- `.get("key", fallback)` returns the fallback instead of crashing on a missing key.
- `.items()` gives you key/value pairs to loop over.

Every API argument in the notebooks is a dictionary.

````python
thinking = {"type": "adaptive", "display": "summarized"}

print("type is:", thinking["type"])
print("missing key with a fallback:", thinking.get("effort", "(not set)"))

READINGS = {"Sydney": 22.5, "Melbourne": 18.0}
print("Perth:", READINGS.get("Perth", "unknown"))   # the tool-result line in 4_tool_schemas

print()
for key, value in thinking.items():       # loop over pairs
    print(f"  {key:>9} = {value}")
````

## 8. Objects and attributes — dots vs. brackets

This trips up everyone:

| | Access with | Example |
|---|---|---|
| **Dictionary** (data you build) | `["key"]` | `thinking["type"]` |
| **Object** (things the SDK returns) | `.attribute` | `response.usage.output_tokens` |

The API response is an **object**, so you write `block.type`, not `block["type"]`.

Below we build fake stand-ins with the same shape as real API responses, so the rest of this
notebook runs with no API key.

````python
class Block:
    """Stand-in for a content block from the API."""
    def __init__(self, **fields):
        self.__dict__.update(fields)     # turn each keyword into an attribute
    def __repr__(self):
        return f"<{self.type}>"

class Usage:
    def __init__(self, input_tokens, output_tokens):
        self.input_tokens = input_tokens
        self.output_tokens = output_tokens

class Response:
    def __init__(self, content, usage):
        self.content = content
        self.usage = usage

response = Response(
    content=[
        Block(type="thinking", thinking="Let me work through the bags of 6 and 11...",
              signature="abc123signature"),
        Block(type="text", text="The answer is 49."),
    ],
    usage=Usage(input_tokens=56, output_tokens=238),
)

print("content is a list:", response.content)
print("first block type: ", response.content[0].type)      # dot, not bracket
print("chained access:   ", response.usage.output_tokens)
````

`hasattr(obj, "name")` asks whether an attribute exists, without crashing:

````python
tb = response.content[0]
print("has .thinking? ", hasattr(tb, "thinking"))
print("has .data?     ", hasattr(tb, "data"))       # a redacted block would have this
````

## 10. Comparisons and membership

| Operator | Asks |
|---|---|
| `==` | equal? (`=` **assigns**, `==` **compares**) |
| `!=` | not equal? |
| `<` `<=` `>` `>=` | ordering — `<=` is "at most" |
| `in` | is this item inside that collection / substring inside that string? |
| `not in` | the opposite |

````python
print("equal:      ", "BILLING" == "BILLING")
print("not equal:  ", "thinking" != "text")
print("at most:    ", abs(8 - 10) <= 2)          # the calibration tolerance check

# `in` works on lists, dicts (keys), sets, and strings
print("in a list:  ", "high" in ["low", "high", "max"])
print("in a dict:  ", "Sydney" in {"Sydney": 22.5})
print("in a string:", "Ben" in "Ben will roll back the deploy")
print("not in:     ", "Ana" not in "Ben will roll back the deploy")
````

`abs(x)` is the distance from zero, so `abs(judge - human) <= 2` means "within 2 points
either way" — the agreement test in the evals notebook.

## 13. Conditional expressions (the "ternary")

A one-line `if`/`else` that **produces a value**:

```
VALUE_IF_TRUE  if  CONDITION  else  VALUE_IF_FALSE
```

⚠️ The **value comes first**, before the condition. That reads backwards if you know
C-style `condition ? a : b`. Read it as *"give me 10, if they match, otherwise 0."*

Because it's an **expression** it can go anywhere a value goes — inside `return`, an
f-string, or a comprehension. A plain `if` block cannot.

````python
def exact_match_long(reference, actual):
    """The long form."""
    if reference.strip() == actual.strip():
        return 10
    else:
        return 0


def exact_match(reference, actual):
    """The same thing as a conditional expression — this is the notebook version."""
    return 10 if reference.strip() == actual.strip() else 0


print(exact_match_long("BILLING", " BILLING "), exact_match("BILLING", " BILLING "))
print(exact_match_long("BILLING", "TECHNICAL"), exact_match("BILLING", "TECHNICAL"))

# Inline, where a statement can't go:
names = ["thinking", "text"]
print(f"first block is {'reasoning' if names[0] == 'thinking' else 'prose'}")
````

## 14. `+=` and booleans as numbers

- `x += 1` is shorthand for `x = x + 1`. Works for `-=`, `*=`, and on strings/lists too.
- **`True` is 1 and `False` is 0.** So `total += flag` counts how many times a flag was true.

That's the `agree += ok` line in the evals notebook — it looks odd until you know that
booleans *are* numbers in Python.

````python
print("True + True =", True + True)      # 2 — booleans really are numbers
print("sum of flags:", sum([True, False, True]))

agree = 0
for judged, human in [(10, 10), (1, 2), (5, 5)]:
    ok = abs(judged - human) <= 2        # a boolean
    agree += ok                          # adds 1 when True, 0 when False
    print(f"  judge={judged:>2} human={human:>2} -> {ok}")

print("agreement:", agree, "of 3")
````

## 15. Comprehensions — build a collection in one line

`[ WHAT_TO_KEEP for ITEM in SEQUENCE if CONDITION ]`

````python
# The long way:
types = []
for block in response.content:
    types.append(block.type)
print("long way: ", types)

# As a comprehension:
print("short way:", [block.type for block in response.content])

# With a filter:
print("filtered: ", [b.text for b in response.content if b.type == "text"])
````

### Dict comprehensions

Same idea with `{key: value for ...}` — builds a dictionary instead of a list.

````python
counts = {name: len(name) for name in ["low", "medium", "xhigh"]}
print(counts)

# Filtering works here too
long_only = {k: v for k, v in counts.items() if v > 3}
print(long_only)
````

## 16. Generators and `next()` — the "find the first match" idiom

Swap `[ ]` for `( )` and you get a **generator**: it produces items one at a time, on demand,
instead of building the whole list. `next(...)` pulls **one** item out and stops looking.

````python
gen = (b for b in response.content if b.type == "thinking")
print("the generator itself:", gen)
print("first match:", next(gen))
````

````python
# As one line — the exact line from the extended thinking notebook:
thinking_block = next(b for b in response.content if b.type == "thinking")

print("found:    ", thinking_block)
print("its text: ", thinking_block.thinking)
print("signature:", thinking_block.signature)
````

**Why not `response.content[0]`?** Position isn't guaranteed. Selecting by `.type` says what
you mean and survives a changed block order.

**Why a generator, not a list?** `next()` stops at the first hit. A list comprehension would
build every match then throw all but one away.

### ⚠️ `next()` crashes when nothing matches

Add a **second argument** for a fallback. Note the extra parentheses: optional when the
generator is the only argument, **required** once you add a default.

````python
no_thinking = [Block(type="text", text="just an answer")]

try:
    next(b for b in no_thinking if b.type == "thinking")
except StopIteration:
    print("crashed with StopIteration")

found = next((b for b in no_thinking if b.type == "thinking"), None)
print("with fallback:", found)

answer = next((b.text for b in no_thinking if b.type == "text"), "")
print("answer:", repr(answer))
````

## 17. Functions

- `def name(inputs):` then an indented body. `return` hands a value back.
- The `"""..."""` first line is a **docstring**.
- `param=value` makes an input **optional**.

````python
def cost(usage, input_rate=5.00, output_rate=25.00):
    """Work out the dollar cost of one response."""
    return (usage.input_tokens * input_rate
            + usage.output_tokens * output_rate) / 1_000_000


print(cost(response.usage))                       # uses the defaults
print(cost(response.usage, output_rate=50.00))    # override just one
print(f"formatted: ${cost(response.usage):.4f}")
````

`1_000_000` is just `1000000` — the underscores are ignored and only there for humans.

### Functions are values, and `lambda` makes a small one inline

A function name without `()` is the function *itself*, so you can pass it to another
function. `lambda args: expression` builds a tiny unnamed one on the spot.

The evals notebook uses this to hand a different grader to the same measuring function.

````python
def measure(score_fn, cases):
    """Apply whatever scoring function it is given."""
    return [score_fn(c) for c in cases]


CASES = [{"human": 10, "judge": 9}, {"human": 2, "judge": 1}, {"human": 5, "judge": 2}]

print("passing a named function:", measure(lambda c: c["judge"], CASES))
print("a different one inline:  ", measure(lambda c: abs(c["judge"] - c["human"]) <= 2, CASES))
````

## 18. Keyword arguments and multi-line calls

`name=value` labels each input, so order stops mattering. Every API call in the notebooks is
written this way, one argument per line, with a **trailing comma** after the last.

````python
def describe(model, max_tokens, thinking=None):
    return f"{model} | max_tokens={max_tokens} | thinking={thinking}"


print(describe(
    model="claude-opus-5",
    max_tokens=4000,
    thinking={"type": "adaptive"},   # <- trailing comma is legal and keeps diffs clean
))

print(describe(max_tokens=100, model="claude-sonnet-5"))   # order doesn't matter
````

## 19. Copying a dictionary with one field changed

Two spellings of the same idea. Both make a **new** dictionary — the original is untouched.

| Form | Notes |
|---|---|
| `{**original, "key": new}` | "unpack everything, then override". Keys can be any string |
| `dict(original, key=new)` | only works when the key is a valid Python name |

`4_tool_schemas` uses `dict(r, tool_use_id=...)` to build a deliberately broken request;
`2_evals_and_judges` uses `{**r, "judge": verdict}` to attach a score to an existing result.

````python
result = {"output": "BILLING", "score": 10}

with_judge = {**result, "judge": 9}          # add a field
overridden = {**result, "score": 0}          # change a field
via_dict   = dict(result, score=0)           # same thing, keyword style

print("original: ", result)                  # unchanged
print("added:    ", with_judge)
print("changed:  ", overridden)
print("dict():   ", via_dict)
````

## 21. `json` — text in, Python out

The API returns **text**. `json.loads(text)` turns that text into Python dicts/lists;
`json.dumps(obj)` goes the other way, and `indent=2` pretty-prints it.

⚠️ **loads = load-**s**tring** (text → Python). **dumps = dump-**s**tring** (Python → text).
The trailing `s` is "string", not plural.

````python
import json

raw = '{"label": "BILLING", "confidence": 0.98}'

parsed = json.loads(raw)                 # text -> Python dict
print("type:", type(parsed).__name__)
print("read a field:", parsed["label"])

print()
print(json.dumps(parsed, indent=2))      # Python -> pretty text
````

## 22. `Counter` — tallying results

`Counter` counts how often each value appears. `4_tool_schemas` uses it to tally which tool
Claude picked across repeated runs.

⚠️ **Two spellings of the same thing** — this is section 9's two import forms in practice, and
the notebooks use both:

| Import | Call it as |
|---|---|
| `import collections` | `collections.Counter(picks)` |
| `from collections import Counter` | `Counter(picks)` |

````python
import collections
from collections import Counter

picks = ["search_knowledge_base", "get_cached_result", "search_knowledge_base",
         "(no tool)", "search_knowledge_base"]

# Identical results — only the import form differs.
tally = Counter(picks)
same  = collections.Counter(picks)
print("from-import form:", tally)
print("module form:     ", same)
print("equal?", tally == same)
print("as a plain dict:", dict(tally))
print("most common:    ", tally.most_common(1))
````

## 23. `try` / `except`

`try:` the risky thing; `except SomeError as e:` runs only if that error happened, with the
error available as `e`. The notebooks use it to *demonstrate* expected API errors without
killing the run.

````python
def risky(n):
    if n == 0:
        raise ValueError("n must not be zero")
    return 100 / n


for n in [4, 0]:
    try:
        print(f"100/{n} =", risky(n))
    except ValueError as e:
        print(f"caught an error: {e}")

print("the notebook keeps running")
````

## 24. Putting it together

A real function from `3_extended_thinking.ipynb`, annotated — running against our fake
`response`, so no API key.

````python
import time

def measure_one(effort, prompt):               # 17. function
    """Run one request at a given effort and report what it cost."""   # docstring
    started = time.time()

    result = response                          # 8. an object with attributes

    answer = next(                             # 16. first match, with a fallback
        (b.text for b in result.content if b.type == "text"),
        "",
    )
    return {                                   # 6. returns a dictionary
        "effort": effort,
        "output_tokens": result.usage.output_tokens,        # 8. chained attributes
        "seconds": round(time.time() - started, 1),         # 20. builtin
        "answer": answer.strip().replace("\n", " ")[:80],   # 2. string methods + slice
    }


results = [measure_one(e, "the puzzle") for e in ["low", "high", "max"]]   # 15. comprehension

for r in results:                              # 11. loop
    # 3. f-string with alignment specs
    print(f"{r['effort']:>5}  {r['output_tokens']:>6} out tok  {r['seconds']:>5}s  {r['answer']}")
````
